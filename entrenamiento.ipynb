```python
import numpy as np
import matplotlib.pyplot as plt


# Datos de la compuerta AND
entradas = np.array([[0,0],[0,1],[1,0],[1,1]])
salidas = np.array([0,0,0,1])


# Función sigmoide
def sigmoide(z):
    return 1 / (1 + np.exp(-z))


# Entrenamiento con gradiente descendente
def entrenar(entradas, salidas, lr=0.1, epochs=10000):
    # Inicialización de parámetros
    peso_entrada1, peso_entrada2, sesgo = np.random.randn(), np.random.randn(), np.random.randn()
    costos = []

    for epoch in range(epochs):
        # Predicción
        z = peso_entrada1*entradas[:,0] + peso_entrada2*entradas[:,1] + sesgo
        predicciones = sigmoide(z)

        # Gradientes
        error = predicciones - salidas
        grad_w1 = np.mean(error * entradas[:,0])
        grad_w2 = np.mean(error * entradas[:,1])
        grad_sesgo = np.mean(error)

        # Actualización de parámetros
        peso_entrada1 -= lr * grad_w1
        peso_entrada2 -= lr * grad_w2
        sesgo -= lr * grad_sesgo

        # Guardar costo
        costo = np.mean((salidas - predicciones)**2)
        costos.append(costo)

        # Ocasionalmente imprimir
        if epoch % 1000 == 0:
            print(f"Epoch {epoch}, Costo: {costo}")

    return peso_entrada1, peso_entrada2, sesgo, costos


# Entrenar la neurona
peso_entrada1, peso_entrada2, sesgo, costos = entrenar(entradas, salidas)
print(f"Pesos entrenados: peso_entrada1={peso_entrada1}, peso_entrada2={peso_entrada2}, sesgo={sesgo}")


# Graficar evolución del costo
plt.plot(costos)
plt.xlabel("Épocas")
plt.ylabel("Costo (MSE)")
plt.title("Evolución del costo durante el entrenamiento")
plt.show()
```
